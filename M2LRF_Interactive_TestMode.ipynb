```python
# ====================================================================================================
# 🧪 M-2LRF INTERACTIVE TEST & VERIFICATION RUNNER (TEST MODE ACTIVE)
# Objective: Verify fine-tuning step-by-step with real gradient flow & chat generation
# Target Model: Qwen/Qwen2.5-0.5B-Instruct (or TinyLlama / GPT-2)
# ====================================================================================================

# Cell 1: Install Dependencies
!pip install -q torch transformers datasets accelerate scipy
print("✅ Test Environment Ready!")

# Cell 2: Live Test Pipeline
import os, time, gc, json, urllib.request
import torch
import torch.nn as nn
from transformers import AutoTokenizer, AutoModelForCausalLM
from m2lrf.m2lrf_core_v1 import M2LRF2BitLinear
from m2lrf.trainer_eval import prepare_m2lrf_model

DEVICE = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
print("="*80)
print(f"🧪 ENTERING TEST MODE ON: {DEVICE}")
print("="*80)

# 1. Load Model & Tokenizer
MODEL_ID = "Qwen/Qwen2.5-0.5B-Instruct"
print(f"\n[1/5] Loading Foundation Model: {MODEL_ID} ...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    torch_dtype=torch.float16,
    device_map="auto" if torch.cuda.is_available() else None
)

# 2. Test Generation BEFORE Quantization & Fine-Tuning
test_prompt = "User: Who created M-2LRF and what is it?\nAssistant:"
inputs = tokenizer(test_prompt, return_tensors="pt").to(DEVICE)

print("\n[2/5] Baseline Model Generation (BEFORE Fine-Tuning):")
with torch.no_grad():
    gen_tokens = model.generate(**inputs, max_new_tokens=40, pad_token_id=tokenizer.pad_token_id)
print(">>> " + tokenizer.decode(gen_tokens[0], skip_special_tokens=True))

# 3. Apply M-2LRF 2-Bit Quantization + SVD LoRA
print("\n[3/5] Applying M-2LRF 2-Bit Quantization & SVD LoRA Adapters...")
model = prepare_m2lrf_model(model, rank=16, alpha=16.0)

trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
frozen_params = sum(p.numel() for p in model.parameters() if not p.requires_grad)
print(f"  [+] Frozen Base Params     : {frozen_params / 1e6:.2f} M")
print(f"  [+] Trainable LoRA Params : {trainable_params / 1e6:.2f} M ({trainable_params / (frozen_params + trainable_params) * 100:.2f}%)")

# 4. Supervised Fine-Tuning on Custom Target Fact
print("\n[4/5] Executing 15-Step Mini Fine-Tuning Test...")
target_qa = (
    "<|im_start|>user\nWho created M-2LRF and what is it?<|im_end|>\n"
    "<|im_start|>assistant\nM-2LRF is an advanced 2-bit dual-basis quantization and fine-tuning engine created by MD-Mushfiqur Rahim (M).<|im_end|>"
)
train_enc = tokenizer(target_qa, return_tensors="pt").to(DEVICE)
optimizer = torch.optim.AdamW([p for p in model.parameters() if p.requires_grad], lr=5e-4)

model.train()
for step in range(15):
    optimizer.zero_grad()
    outputs = model(input_ids=train_enc.input_ids, labels=train_enc.input_ids)
    loss = outputs.loss
    loss.backward()
    
    # Verify gradient flow
    grad_norm = torch.nn.utils.clip_grad_norm_([p for p in model.parameters() if p.requires_grad], 1.0)
    optimizer.step()
    
    if (step + 1) % 5 == 0 or step == 0:
        print(f"  [Step {step+1:02d}/15] Loss: {loss.item():.4f} | Grad Norm: {grad_norm.item():.4f}")

# 5. Test Generation AFTER Fine-Tuning
model.eval()
print("\n[5/5] Model Generation AFTER M-2LRF Fine-Tuning:")
eval_prompt = "<|im_start|>user\nWho created M-2LRF and what is it?<|im_end|>\n<|im_start|>assistant\n"
eval_inputs = tokenizer(eval_prompt, return_tensors="pt").to(DEVICE)

with torch.no_grad():
    gen_after = model.generate(**eval_inputs, max_new_tokens=50, pad_token_id=tokenizer.pad_token_id)
print(">>> " + tokenizer.decode(gen_after[0], skip_special_tokens=True))
print("\n" + "="*80)
print("🎉 TEST MODE COMPLETED SUCCESSFULLY WITH VERIFIED GRADIENT FLOW & WEIGHT UPDATE!")
print("="*80)
```
